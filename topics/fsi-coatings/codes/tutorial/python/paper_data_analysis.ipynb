# 论文数据分析与可视化

> 戚彦福(2026) 钢桥涂层的老化机理与耐久性研究

根据论文中的典型数据, 复现关键图表

```python
import numpy as np
import matplotlib.pyplot as plt
plt.rcParams.update({'font.size':11,'figure.dpi':150})
```

## 附着力时变曲线 (第3章)

```python
t = np.linspace(0,200,10)
A_25C = 100*np.exp(-0.0035*t)
A_50C = 100*np.exp(-0.008*t)
A_70C = 100*np.exp(-0.016*t)

plt.figure(figsize=(8,4))
plt.plot(t,A_25C,'o-',label='25°C')
plt.plot(t,A_50C,'s-',label='50°C')
plt.plot(t,A_70C,'^-',label='70°C')
plt.xlabel('老化时间 (天)');plt.ylabel('附着力 (%)')
plt.grid(alpha=0.3);plt.legend()
plt.tight_layout();plt.savefig('adhesion_curves.png',dpi=150)
plt.show()
```

## Arrhenius 分析 (第4章)

```python
# 不同温度下的反应速率
T_K = np.array([298, 323, 343])
k = np.array([0.0035, 0.008, 0.016])
R = 8.314
# 拟合
coeffs = np.polyfit(1/T_K, np.log(k), 1)
U0 = -coeffs[0]*R

plt.figure(figsize=(8,4))
plt.plot(1/T_K*1000, np.log(k), 'o', markersize=8)
plt.plot(1/T_K*1000, np.polyval(coeffs, 1/T_K), '-', label=f'U0={U0/1000:.1f} kJ/mol')
plt.xlabel('1000/T (1/K)');plt.ylabel('ln(k)')
plt.grid(alpha=0.3);plt.legend()
plt.tight_layout();plt.show()
```

## 寿命预测柱状图 (第6章)

```python
systems = ['普通型','长效型','水性体系']
life_static = [12, 22, 25]
life_dynamic = [14, 31, 31]
spec = [30, 30, 30]

x = np.arange(len(systems))
w = 0.25
plt.figure(figsize=(8,4))
plt.bar(x-w, life_static, w, label='静应力', color='#EE7733')
plt.bar(x, life_dynamic, w, label='动应力', color='#0077BB')
plt.bar(x+w, spec, w, label='规范要求30年', color='#CCCCCC', alpha=0.5)
plt.axhline(y=30, color='r', linestyle='--', alpha=0.5)
plt.xticks(x, systems);plt.ylabel('寿命 (年)')
plt.legend();plt.grid(alpha=0.3,axis='y')
plt.tight_layout();plt.savefig('life_prediction.png',dpi=150)
plt.show()
```
